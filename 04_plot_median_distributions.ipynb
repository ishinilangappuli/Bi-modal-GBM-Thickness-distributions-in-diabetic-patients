```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load full component data
CSV_PATH = r"C:\Users\ishin\OneDrive\Desktop\ish\gbm_thickness_results.csv"
df = pd.read_csv(CSV_PATH)

# Set clean aesthetic style
sns.set_theme(style="whitegrid")
patient_order = sorted(df['patient_id'].unique())

plt.figure(figsize=(12, 6))

# Boxplot comparing medians and interquartile ranges (Warning fixed)
sns.boxplot(
    data=df, 
    x='patient_id', 
    y='median_thickness_nm', 
    hue='patient_id',       # Assign hue to x variable
    legend=False,           # Disable redundant legend
    order=patient_order, 
    palette="Set3", 
    boxprops=dict(alpha=0.8)
)

plt.axhline(y=350, color='red', linestyle='--', linewidth=1.5, label='Normal Reference (~350 nm)')

plt.title("GBM Median Thickness per Component across Patients", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Patient ID", fontsize=12, fontweight='bold')
plt.ylabel("Median Thickness (nm)", fontsize=12, fontweight='bold')
plt.xticks(rotation=45)
plt.legend(loc="upper right")
plt.tight_layout()
plt.savefig(r"C:\Users\ishin\OneDrive\Desktop\ish\gbm_median_boxplot.png", dpi=300)
plt.show()
```
