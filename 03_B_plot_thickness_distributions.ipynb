```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load results from Steps 2 & 3
CSV_PATH = r"C:\Users\ishin\OneDrive\Desktop\ish\gbm_thickness_results.csv"
df = pd.read_csv(CSV_PATH)

# Set clean aesthetic style
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams.update({'font.size': 11})

# Sort patient IDs numerically/alphabetically
patient_order = sorted(df['patient_id'].unique())

# ==========================================
# 1. BOXPLOT WITH OVERLAID DATA POINTS
# ==========================================
plt.figure(figsize=(12, 6))
sns.boxplot(
    data=df, x='patient_id', y='mean_thickness_nm', 
    order=patient_order, color="lightskyblue", boxprops=dict(alpha=0.6)
)
sns.stripplot(
    data=df, x='patient_id', y='mean_thickness_nm', 
    order=patient_order, color="navy", alpha=0.7, jitter=0.2, size=5
)

# Reference line for normal adult mean thickness (~350 nm)
plt.axhline(y=350, color='red', linestyle='--', linewidth=1.5, label='Normal Baseline (~350 nm)')

plt.title("GBM Mean Thickness Distribution per Patient", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Patient ID", fontsize=12, fontweight='bold')
plt.ylabel("Mean Thickness (nm)", fontsize=12, fontweight='bold')
plt.xticks(rotation=45)
plt.legend(loc="upper right")
plt.tight_layout()
plt.savefig(r"C:\Users\ishin\OneDrive\Desktop\ish\gbm_boxplot_per_patient.png", dpi=300)
plt.show()

# ==========================================
# 2. KDE DENSITY PLOT (FOR MULTIMODALITY CHECK)
# ==========================================
plt.figure(figsize=(10, 6))
sns.kdeplot(
    data=df, x='mean_thickness_nm', hue='patient_id', 
    common_norm=False, fill=False, linewidth=1.5, alpha=0.8
)
plt.axvline(x=350, color='red', linestyle='--', linewidth=1.5, label='Normal Baseline (~350 nm)')

plt.title("GBM Thickness Density Curves (Checking for Multimodal Peaks)", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Mean Thickness (nm)", fontsize=12, fontweight='bold')
plt.ylabel("Density", fontsize=12, fontweight='bold')
plt.legend(title="Patient ID", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.savefig(r"C:\Users\ishin\OneDrive\Desktop\ish\gbm_kde_distribution.png", dpi=300)
plt.show()
```
