```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

CSV_PATH = r"C:\Users\ishin\OneDrive\Desktop\ish\gbm_thickness_results.csv"
df = pd.read_csv(CSV_PATH)

# Select representative patient with distinct peaks (e.g., '07-25')
target_patient = '07-25'
data = df[df['patient_id'] == target_patient]['median_thickness_nm'].dropna().values.reshape(-1, 1)

# Fit GMM with 2 components
gmm = GaussianMixture(n_components=2, random_state=42).fit(data)
means = gmm.means_.flatten()
covariances = gmm.covariances_.flatten()
weights = gmm.weights_.flatten()

# Plotting setup
x = np.linspace(data.min() - 50, data.max() + 50, 1000).reshape(-1, 1)
plt.figure(figsize=(9, 5))

# Plot actual data histogram
plt.hist(data, bins=15, density=True, alpha=0.4, color='skyblue', edgecolor='black', label='Observed Data')

# Plot individual GMM Gaussians
for i in range(2):
    mean = means[i]
    std = np.sqrt(covariances[i])
    weight = weights[i]
    y = weight * (1 / (std * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean) / std) ** 2)
    plt.plot(x, y, label=f'Peak {i+1} Component ($\mu={mean:.1f}$ nm)')

plt.title(f"GMM 2-Component Decomposition for Patient {target_patient}", fontsize=13, fontweight='bold')
plt.xlabel("Median Thickness (nm)", fontsize=11, fontweight='bold')
plt.ylabel("Density", fontsize=11, fontweight='bold')
plt.legend()
plt.tight_layout()
plt.savefig(rf"C:\Users\ishin\OneDrive\Desktop\ish\gmm_fit_{target_patient}.png", dpi=300)
plt.show()
```
